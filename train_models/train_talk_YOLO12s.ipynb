```python
from ultralytics import YOLO
model = YOLO('yolo12s.pt')
```

```python
import kagglehub

# Download latest version
dataset_path = kagglehub.dataset_download("shreyasudaya/scb-05-dataset")

print("Path to dataset files:", dataset_path)
```

```python
import os
from ultralytics import YOLO

# Paths
yaml_path = '/content/SCB5-talk.yaml'
src_yaml = os.path.join(dataset_path, 'SCB-Dataset/SCB5-Talk-2024-9-17/SCB5-Talk-2024-9-17.yaml')

# Fix path inside YAML
with open(src_yaml) as f:
    content = f.read().replace('/data/SCB/SCB-dataset/SCB5-Talk-2024-9-17', os.path.join(dataset_path, 'SCB-Dataset/SCB5-Talk-2024-9-17'))

with open(yaml_path, 'w') as f:
    f.write(content)

# Train
results = model.train(data=yaml_path, epochs=100, imgsz=640, batch=8, project='student_behavior1', name='talk-model')
```
